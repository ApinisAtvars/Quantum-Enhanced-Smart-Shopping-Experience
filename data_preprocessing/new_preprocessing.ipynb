# Preprocessing for existing NCF model
>Just involves renaming the columns, no negative sampling even for implicit feedback

>Everything else handled by data.py

### 1. Read CSV

```python
import pandas as pd
```

```python
ratings = pd.read_csv('../data/ml-32m/ratings.csv')
```

```python
ratings.head()
```

### 2. Rename columns

```python
ratings.rename(mapper={'movieId': 'itemId'}, inplace=True, axis=1)
```

```python
ratings.head()
```

```python
import os

os.makedirs('../data/ncf_preprocessed', exist_ok=True)
ratings.to_csv('../data/ncf_preprocessed/ratings.csv', index=False)
```

### 3. Create smaller subsets

```python
# Create a subset of 1000 users
subset = ratings['userId'].drop_duplicates().sample(n=1000, random_state=42)
subset_ratings = ratings[ratings['userId'].isin(subset)]
subset_ratings.to_csv('../data/ncf_preprocessed/ratings_subset.csv', index=False)
```

```python
# Create a subset of 500 users
subset = ratings['userId'].drop_duplicates().sample(n=500, random_state=42)
subset_ratings = ratings[ratings['userId'].isin(subset)]
subset_ratings.to_csv('../data/ncf_preprocessed/ratings_subset_500.csv', index=False)
```
